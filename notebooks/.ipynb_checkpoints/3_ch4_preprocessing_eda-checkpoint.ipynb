# Resample the full dataset

resample('W').mean() divides the timeline into weekly bins and calculates the mean of observations that fall in each bin.  If no observation is in the weekly bin, then NaN is introduced.

The full dataset will be resampled to weekly ('W-TUE') ahead of train/test split and subsequent preprocessing because:
1.  The forecasting models require a regular weekly time index, and the empirical sampling frequency is approximately weekly, with 52 observations per year being the modal annual count in the historical data (see notebook 2_ch4_eda_clean). The full aggregated series is therefore placed on a weekly calendar grid before interpolation.
2.  Resampling weekly as 'W-FRI' - 'W-SUN' pushes the last observation into the next calendar year.  Obviously this will depend on the calendar year for subsequent data additions from NOAA and should be reevaluated for subsequent modeling and forecasting that includes new data.  
3. Resampling the complete series before splitting also ensures that the weekly time grid is defined consistently across the train/test boundary. This avoids boundary inconsistencies encountered when train and test sets were resampled independently.

```python
# Inspect the collection-day distribution to identify a sensible weekly anchor.

df.index.day_name().value_counts()
```

Tuesday is the most common collection day. Using W-TUE also prevents the final 2025 observations (2025-12-30) from being labeled as a 2026-01-04 weekly endpoint under the default W-SUN convention.

```python
# observations land on a Tuesday with the highest frequency.  
# resample with weekly frequency with a Tuesday anchor

df_float = df[['value', 'value_unc']].resample('W-TUE').mean()
df_obj = (
    df[['qcflag']]
        .resample('W-TUE')
        .agg(lambda x: x.mode().iloc[0] if not x.mode().empty else np.nan)
)

df_resamp = pd.concat([df_float, df_obj], axis=1)
```

**Interpretation of preprocessed CH4 data**

*stationarity*: The preprocessed cleaned CH4 data series is non-stationary according to the ADF and KPSS tests.  The ADF test fails to reject the unit-root null hypothesis (p = 0.914), while the KPSS test rejects the null hypothesis of stationarity (p = 0.010). Combined, this strongly suggests that the resampled, preprocessed data has a stochastic trend.  Also, and importantly, this result is consistent with the pronounced long-term positive trend identified in the EDA of the annual mean concentrations (slope = 5.66 ppb/year, see 2_ch4_eda_clean notebook).  So, preprocessing hasn't substantially altered the general trend of the data.  

*scedasticity*: Breusch–Pagan and White tests applied to the STL residuals found no evidence of heteroscedasticity (p = 0.578 and p = 0.841, respectively). Thus, after removing the estimated trend and seasonal components, the residual variance appears reasonably stable indicating that an additional variance-stabilizing transformation does not appear necessary based on these diagnostics.

*autocorrelation*: The ACF and PACF of the STL residuals suggest both short-term and annual-scale dependence.
- At short lags, the ACF and PACF show a prominent lag-1 relationship, with substantially weaker correlations at subsequent short lags. This indicates short-term dependence but does not, by itself, distinguish among AR, MA, and ARMA structures.
- A significant negative spike at approximately lag 52 indicates that annual-scale dependence remains in the residuals after STL decomposition. The atmospheric CH₄ seasonal cycle is not perfectly fixed from year to year, so STL may not remove all annual-scale variation.
- A borderline significant PACF value at lag 104 indicates possible conditional dependence at approximately two seasonal periods. This provides additional evidence of multi-year seasonal structure, but the ACF/PACF pattern alone is insufficient to determine the appropriate seasonal AR or MA order.

*ACF/PACF caveat*: No median smoothing is applied by the GasPreprocessor since the smoothing window = 1.  However, interpolation can affect short-lag autocorrelation if there are enough consecutive missing/outlier observations, so it isn't possible to claim that the lag-1 structure is necessarily entirely natural atmospheric persistence. A safe to interpretation is that the short-lag autocorrelation is structure remaining after preprocessing and STL decomposition, without attributing it to median smoothing. The ACF/PACF diagnostics are used to generate candidate SARIMA specifications rather than to determine the final model order.

- *Candidate starting models*: The short-term lag-1 dependence and remaining annual-scale dependence motivate testing SARIMA models containing both non-seasonal and seasonal terms. Initial candidates include:
    - SARIMA(1,1,1)(1,0,0,52)
    - SARIMA(0,1,0)(1,0,0,52)

These are exploratory starting specifications rather than final model selections. The need for non-seasonal and seasonal differencing, alternative AR/MA orders, residual diagnostics, and out-of-sample forecasting performance will be evaluated systematically in Notebook 4.

**Next steps**:
- assess the need for non-seasonal and seasonal differencing
- determine if the seasonal cycles are additive or multiplicative
- examine the ACF and PACF plots of the first-differenced and seasonally-differenced data to estimate the non-seasonal (p,q) and seasonal (P,Q) SARIMA orders. The differencing analysis will be done in the next notebook (4_ch4_modeling).

## Trend slope

```python
# calculate the yearly average
yearly_avg_df = ch4_preprocessed.groupby(ch4_preprocessed.index.year).mean()

# calculate the slope
gas_trend_data = []

x = yearly_avg_df.index
y = yearly_avg_df

# linear regression
slope, intercept, r_value, p_value, std_err = linregress(x,y)

# predicted line
y_pred = intercept + slope * x

# confidence intervals
ci_lower = slope - 1.96 * std_err
ci_upper = slope + 1.96 * std_err

gas_trend_data_df = pd.DataFrame(gas_trend_data)

# plot the data
plt.figure(figsize=(12,6))
plt.plot(x, y, 'o-', label='Yearly mean CH4')
plt.plot(x, y_pred, 'r--', label='Linear trend')
plt.title("Yearly Averaged 'CH4_preprocessed' Data Over Time", fontsize=18)
plt.ylabel('[CH4] ppb', fontsize=14)
plt.xlabel('Year', fontsize=14)
plt.yticks(fontsize=12)
plt.xticks(fontsize=12)

# metrics text box
textstr = '\n'.join((
    f'Slope = {slope:.3f} +/- {std_err:.3f} ppb/yr',
    f'95% CI = [{ci_lower:.3f}, {ci_upper:.3f}]',
    f'R^2 = {r_value**2:.3f}',
    f'p = {p_value:.2e}'
))

# place box on the figure
plt.gca().text(
    0.02, 0.95, textstr, transform=plt.gca().transAxes,
    fontsize=14, verticalalignment='top',
    bbox=dict(boxstyle='square,pad=0.4', facecolor='white', edgecolor='none')
)

plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(fontsize=14)
plt.tight_layout()
plt.show()
```

- **Slope:**  CH4 concentration rose by 5.68 ppb/year over the ~35 years of data analyzed here [1,2].
- **CI:** The CI (confidence interval) values suggest that the "true" annual increase is somewhere between 5.1 and 6.3 ppb/year.
- **R^2:** The R^2 value is strong and indicates that 91.7% of yearly variation is explained by the linear trend.
- **p:** The p-value indicates the probability of getting this slope by chance, which is extremely low.  

The plot of yearly averages show a stabiliation between approximately 2000 - 2007 [3] and then an increase in rate around 2007 [4,5].

## Non-seasonal differencing test

```python
# estimate the need for and number of non-seasonal with pmdarima "ndiffs"
# Purpose: Determines the d parameter for an ARIMA(p, d, q) model by removing primary trends, 
# random walks, or shifts in the mean over time.

d = ndiffs(ch4_preprocessed, test='kpss')
print(f'Estimated non-seasonal differences: d = {d}')
```

```python
# Verify that d=1 is the correct degree of differencing (minimum variance principle)

# difference the series
ndiff1 = ch4_preprocessed.diff().dropna()
ndiff2 = ndiff1.diff().dropna()

# determine the variance of each series: d=0, d=1, d=2
var_d0 = np.var(ch4_preprocessed)
var_d1 = np.var(nsdiff1)
var_d2 = np.var(nsdiff2)

print(f'variance at d=0: {var_d0:.3f}')
print(f'variance at d=1: {var_d1:.3f}')
print(f'variance at d=2: {var_d2:.3f}')
```

**Interpretation of the ndiffs (KPSS) test**

- The ndiffs test returns d = 1 for the need and degree of differencing.
- The variance results confirm that d = 1 is the correct degree of differencing based on the minimal variance principle.  

# Seasonal structure: shape, amplitude, and transformation considerations

## Seasonal differencing test

*Question*: Does the seasonal unit-root test indicate that seasonal differencing is needed for the weekly CH4 series?

*Purpose*: Provides an initial estimate of the seasonal differencing order (D) for a SARIMA model with seasonal period (m=52).

```python
# Run the OCSB (Osborn, Chui, Smith, and Birchenhall) test to statistically determine if seasonal differencing is required

# estimate seasonal differences using OCSB test automatically
D = nsdiffs(ch4_preprocessed, m=52, test='ocsb')

print(f'Seasonal differences needed: D = {D}')
```

```python
# The global trend is much more substantial than the amplitudes of the seasonal cycles.

# estimate seasonal differences using OCSB test where d=1
D = nsdiffs(ndiff1, m=52, test='ocsb')

print(f'Seasonal differences needed (D): {D}')
```

**Interprtetation of the OCSB (nsdiffs) test**: 

The OCSB test returned (D=0) for both the preprocessed CH4 series and the first-differenced series. The test therefore does not recommend seasonal differencing under the specifications examined. This does not establish that seasonality is deterministic or unchanging. The stability of seasonal shape and amplitude is examined separately below.

Because this is an exploratory analysis of the full series, the final differencing decisions will be made using the training data in the subsequent modeling notebook.

## Inspect seasonal structure shape

```python
# check how many weekly observations each year contributes
# the data was resampled to 'W-TUE' frequency.
# some years will have 52 or 53 Tuesdays.  

seasonal_component = stl_result.seasonal
weeks_per_year = seasonal_component.groupby(seasonal_component.index.year).count()

print(f"Min weeks per year: {weeks_per_year.min()}")
print(f"Max weeks per year: {weeks_per_year.max()}")

print(f"Number of years with 52 Tuesdays: {(weeks_per_year == 52).sum()}")
print(f"Number of years with 53 Tuesdays: {(weeks_per_year == 53).sum()}")
```

Since some years have 53 datapoints, I will create a dictionary of the seasonal position for each year. The seasonal position is a fractional position of the indexed day of the year / days in the given year. 


process for creating seasonal positions:
1. Extract the STL seasonal component.
2. Assign each observation to a calendar year.
3. Convert each date to fractional position within that year.
4. Interpolate every year onto the same 52-point phase grid (because some years have 53 datapoints).

```python
# 1. Extract the STL seasonal component
# put the seasonal component into a DataFrame
seasonal_component.name = 'seasonal'
seasonal_df = seasonal_component.to_frame()

# 2. Assign each observation to a calendar year.
seasonal_df['year'] = seasonal_df.index.year

# 3. Convert each date to fractional position within that year.
# dynamically find the total number of days in each specific calendar year to 
# handle 365-day normal years and 366-day leap years.

# Number of days in each calendar year
days_in_year = np.where(
    seasonal_df.index.is_leap_year,
    366,
    365
)

# Fractional position within the calendar year
seasonal_df['phase'] = (
    (seasonal_df.index.dayofyear - 1) / days_in_year
)

# 4. Interpolate every year onto the same 52 point phase grid 
# because some years have 53 datapoints

# common annual grid representing 52 seasonal positions
phase_grid = np.linspace(0,1,52, endpoint=False)

yearly_curves = {}

for year, group in seasonal_df.groupby('year'):
    group = group.sort_values('phase')

    # interpolate that year's weekly observations onto the common 52 position annual grid
    yearly_curves[year] = np.interp(
        phase_grid,
        group['phase'],
        group['seasonal']
    )

yearly_curves = pd.DataFrame(yearly_curves).T

yearly_curves.index.name = 'year'
yearly_curves.columns = np.arange(1,53)
yearly_curves.columns.name = 'seasonal_position'

# Calculate the mean seasonal curve for comparison 
mean_seasonal_curve = yearly_curves.mean(axis=0)
```

```python
# plot all curves

#calculate the mean of all yearly curves
mean_seasonal_curve = yearly_curves.mean(axis=0)

# year range for plot title
series = ch4_preprocessed.index.year
first_year = series[0]
last_year = series[-1]

plt.figure(figsize=(14,6))

for year in yearly_curves.index:
    plt.plot(
        yearly_curves.columns,
        yearly_curves.loc[year],
        alpha=0.4,
        linewidth=1
    )

# plot the mean seasonal curve
plt.plot(
    yearly_curves.columns,
    mean_seasonal_curve,
    linewidth=3,
    label='Mean Seasonal Curve'
)

plt.axhline(0, linewidth=1)
plt.tick_params(axis='both', labelsize=14)
plt.xlabel('Position Within Annual Cycle', fontsize=16)
plt.ylabel('STL Seasonal Component (ppb)', fontsize=16)
plt.legend(loc='lower left', fontsize=16)
plt.title(f"Yearly Aligned STL Seasonal Components ({first_year}-{last_year})", fontsize=18)

plt.tight_layout()
plt.show()
```

**Observations of the yearly seasonal shapes**

The seasonal shape is remarkably consistent over the 35 years of data.  However, there is substantial variation in the intra-seasonal values at seaonal positions over the years.

## Inspect seasonal amplitude

```python
# get the stl_results.  
# stl_result is stored as an instance attribute
stl_result = preprocessor.stl_result_

# Group raw data by year to compute the annual mean
trend_component = stl_result.trend
yearly_mean = trend_component.groupby(trend_component.index.year).mean()

# Extract yearly amplitudes from the STL decomposition result
seasonal_component = stl_result.seasonal
yearly_amplitudes = seasonal_component.groupby(seasonal_component.index.year).agg(lambda x: x.max() - x.min())

# Calculate the single average seasonal amplitude for the legend
avg_amp = yearly_amplitudes.mean()

# Align the data (ensures both series match the exact same years, drop missing values)
combined = pd.DataFrame({'mean': yearly_mean, 'amplitude': yearly_amplitudes}).dropna()

# Extract x and y as 1D arrays for the regression line
x = combined.index.to_numpy()
y = combined['amplitude'].to_numpy()

# Calculate average amplitude for the years included
avg_amp = combined['amplitude'].mean()

# plot yearly amplitude over time
fig, ax = plt.subplots(figsize=(10,4))

plt.plot(x,y, marker='o')

ax.axhline(
    avg_amp,
    color='gray',
    linestyle='--',
    label=f"Average amplitude = {avg_amp:.2f} ppb"
)
    
ax.set_xlabel('Year', fontsize=14)
ax.set_ylabel('Seasonal Amplitude (ppb)', fontsize=14)
ax.set_title(f"Annual CH4 Seasonal Amplitude ({x.min()}–{x.max()})", fontsize=18)
ax.grid(True, linestyle='--', alpha=0.5)
ax.legend(fontsize=12)

plt.show()
```

```python
# further analysis of 1991 and the apparent trend of increasing amplitude vs mean level for 2021-2025.
# are the extreme values (troughs and peaks) reliable?  
# inspect the location of the trough and peak values relative to the actual observation. 

# Build original CH4 observation series indexed by weekly date.
# df_resamp contains the data BEFORE GasPreprocessor interpolation.

weekly_ch4 = df_resamp[['date', 'CH4']].copy()
weekly_ch4['date'] = pd.to_datetime(weekly_ch4['date'])

observed_ch4 = (
    weekly_ch4
    .set_index('date')['CH4']
    .sort_index()
    .reindex(seasonal_component.index)
)

summary_rows = []

for year in sorted(seasonal_component.index.year.unique()):

    # Original weekly STL seasonal component for this calendar year
    s = seasonal_component.loc[
        seasonal_component.index.year == year
    ]

    # Original observation status at those same weekly dates
    obs = observed_ch4.reindex(s.index)

    peak_date = s.idxmax()
    trough_date = s.idxmin()

    observed_dates = obs.index[obs.notna()]

    def nearest_observation_weeks(date):
        if len(observed_dates) == 0:
            return np.nan

        distances = np.abs(
            (observed_dates - date) / pd.Timedelta(weeks=1)
        )

        return int(distances.min())

    summary_rows.append({
        'year': year,
        'grid_weeks': len(s),
        'observed_weeks': int(obs.notna().sum()),
        'coverage_pct': 100 * obs.notna().mean(),

        'peak_date': peak_date,
        'peak_seasonal_ppb': s.loc[peak_date],
        'peak_week_observed': pd.notna(obs.loc[peak_date]),
        'peak_nearest_observation_weeks':
            nearest_observation_weeks(peak_date),

        'trough_date': trough_date,
        'trough_seasonal_ppb': s.loc[trough_date],
        'trough_week_observed': pd.notna(obs.loc[trough_date]),
        'trough_nearest_observation_weeks':
            nearest_observation_weeks(trough_date),
    })

extrema_coverage_df = pd.DataFrame(summary_rows).set_index('year')

# Display the recent five years first
years_to_display = [1991] + list(range(2021, 2026))
extrema_coverage_df.loc[years_to_display].round(2)                
```

Important implications of this information:

**1991**: 1991 has the third largest amplitude of the 35 years analyzed, but also has the lowest correlation with the mean seasonal curve (0.62 vs 0.89).  A concern is that the extreme amplitude and low correlation value could be due to STL decomposition smoother edge effects.  Coverage analaysis reveals that the 1991 peak and trough dates both occur in weeks with an observation.  So, these values are less likely to be influenced by interpolation.  While this brief analysis doesn't rule out STL edge effects, neither does it argue for removing the resampled and preprocessed 1991 data from subsequent model evaluations and forecasting.   

**2021-2025**: The estimated annual CH4 seasonal amplitude increases progressively from 2022 through 2025. Inspection of the original observation coverage shows that the seasonal peaks occur on weeks with direct observations in 2022–2024, while both the peak and trough in 2025 occur on missing weeks but are within one week of an observed measurement. 

Those missing CH4 datapoints are estimated by linear interpolation as values between the surrounding observed measurements. Impotatnly, interpolation cannot directly create values that exceed the upper endpoint or fall below the lower endpoint of a gap, reducing the concern that interpolation itself introduced artificial extremes into the input series. Although interpolation may still influence the estimated amplitudes, particularly in 2025, and STL decomposition can affect the estimated seasonal peaks and troughs, their occurrence on observed weeks in most recent years provides additional, but not conclusive, support for the observed increase in seasonal amplitude.

The observed pattern is physically plausible and consistent with independent research reporting increasing methane seasonal-cycle amplitude at Mauna Loa and other sites over an earlier historical period [1]. Nevertheless, the declining observational coverage in recent years limits confidence in the absolute amplitude estimates. The results therefore suggest a potentially meaningful increase in seasonal amplitude, while leaving its precise magnitude and physical attribution uncertain.

## Inspect robustness of amplitude estimates

Does the STL seasonal smoothing window affect or contribute to the trends in seasonal amplitudes?  

```python
# Use the exact preprocessed series used for the final STL decomposition.
# This does not rerun cleaning, outlier removal, or interpolation.
analysis_series = preprocessor.cleaned_series_.copy()

if analysis_series.isna().any():
    raise ValueError(
        "The preprocessed series contains NaNs; inspect before proceeding."
    )

# Retrieve the trend-smoother window implied by the existing defaults.
# Holding it fixed avoids changing two smoothing parameters simultaneously.
baseline_stl = STL(
    analysis_series,
    period=preprocessor.seasonal_period,
    seasonal=7,
    robust=True
)

trend_window = baseline_stl.config['trend']

print(f"Period: {preprocessor.seasonal_period}")
print(f"Fixed trend-smoother window: {trend_window}")

# Compare the default seasonal smoother with a wider smoother.
seasonal_windows = [3, 7, 13]

stl_results = {}
annual_amplitudes = {}

for seasonal_window in seasonal_windows:

    result = STL(
        analysis_series,
        period=preprocessor.seasonal_period,
        seasonal=seasonal_window,
        trend=trend_window,
        robust=True
    ).fit()

    stl_results[seasonal_window] = result

    annual_amplitudes[seasonal_window] = (
        result.seasonal
        .groupby(result.seasonal.index.year)
        .agg(lambda x: x.max() - x.min())
    )

# Compare annual amplitude estimates
amplitude_sensitivity = pd.DataFrame({
    f'Seasonal smoother = {window}':
        annual_amplitudes[window]
    for window in seasonal_windows
})

amplitude_sensitivity.head()
```

```python
fig, ax = plt.subplots(figsize=(10, 4))

amplitude_sensitivity.plot(
    ax=ax,
    marker='o'
)

ax.set_xlabel('Year', fontsize=14)
ax.set_ylabel('Seasonal Amplitude (ppb)', fontsize=14)
ax.set_title('Sensitivity of Annual CH₄ Amplitude to STL Smoothing', fontsize=18)
ax.grid(True, linestyle='--', alpha=0.5)
ax.legend(loc='upper left', fontsize=10)

plt.show()
```

**Interpretation of the amplitude senstivity plot**

This plot demonstrates that the seaonal amplitude and multiyear pattern is sensitive to the STL smoothing window.  
- smoother = 3 is a short seasonal-smoother window that allows the estimated seasonal pattern to change more rapidly across years, producing pronounced year-to-year variation. This may reflect local structure, measurement gaps, interpolation effects, or some combination; the sensitivity analysis does not isolate their individual contributions.  
- smoother = 7 is the default statsmodels setting.  The series is less jagged and presents what appears to be a multiyear pattern.
- smoother = 13 imposes a greater smoothness on the series seasonal comonents year-to-year change.

*Note*: statsmodels recommends a seasonal smoother window >= 7, which is the default setting.  While window = 3 is the minimal allowed setting, this window is used here purely as a test of seasonal sensitivity, not as an option for parameterization.  

```python
difference = (
    amplitude_sensitivity['Seasonal smoother = 13']
    - amplitude_sensitivity['Seasonal smoother = 7']
)

print(f"Correlation between amplitude estimates: "
      f"{amplitude_sensitivity.corr().iloc[0, 1]:.3f}")

print(f"Mean absolute difference: "
      f"{difference.abs().mean():.2f} ppb")

print(f"Maximum absolute difference: "
      f"{difference.abs().max():.2f} ppb")

display(
    amplitude_sensitivity.loc[2021:2025]
    .assign(difference_13_minus_7=difference)
)
```

**Sensitivity of seasonal amplitude to STL smoothing**

A sensitivity analysis compared annual seasonal amplitudes estimated using STL seasonal-smoother windows of 7 and 13, while holding the other decomposition settings fixed. The annual amplitude estimates were positively correlated (r = 0.818), indicating that the two specifications captured related interannual variation, although the mean absolute difference was 6.38 ppb. Both specifications showed an increase in estimated seasonal amplitude from 2021 through 2025, but the estimated magnitude of this increase was substantially smaller with the wider smoother.

These findings suggest that the broad recent increase is not unique to the default STL setting, but its estimated magnitude is sensitive to the smoothing specification. The observed multiyear amplitude fluctuations may reflect changes in atmospheric processes, decomposition-related smoothing, or both. This analysis does not establish their physical cause or demonstrate a regular oscillation. Annual amplitude estimates are therefore interpreted as descriptive quantities rather than exact measurements of the underlying physical seasonal amplitude.

## Amplitude vs mean CH4 concentration

```python
# plot amplitude of the seasonal cycle vs mean CH4 concentration, with or without first and/or last year (edge cases)

# get the stl_results.  
# stl_result is stored as an instance attribute
stl_result = preprocessor.stl_result_

# Group raw data by year to compute the annual mean
trend_component = stl_result.trend
yearly_mean = trend_component.groupby(trend_component.index.year).mean()

# Extract yearly amplitudes from the STL decomposition result
seasonal_component = stl_result.seasonal
yearly_amplitudes = seasonal_component.groupby(seasonal_component.index.year).agg(lambda x: x.max() - x.min())

# Calculate the single average seasonal amplitude for the legend
# avg_amp = yearly_amplitudes.mean()

# Align the data (ensures both series match the exact same years, drop missing values)
# exclude boundary years weith known edge effects or partial data (see comments above)
boundary_years = [None]  # or enter years to exclude, e.g. [1991, 2025]
obs_per_year = seasonal_component.groupby(seasonal_component.index.year).count()
complete_years = obs_per_year[
    (obs_per_year >= 50) & (~obs_per_year.index.isin(boundary_years))
    ].index

combined = pd.DataFrame({
    'mean'     : yearly_mean,
    'amplitude': yearly_amplitudes
}).loc[complete_years].dropna()

# Extract x and y as 1D arrays for the regression line
x = combined['mean'].values
y = combined['amplitude'].values

# Calculate the regression line
m, b = np.polyfit(x, y, 1)

# Calculate the correlation coefficient (r)
corr = combined['mean'].corr(combined['amplitude'])

# Calculate the single average seasonal amplitude for the legend
avg_amp = yearly_amplitudes[~yearly_amplitudes.index.isin(boundary_years)].mean()

# get index years following boundary year exclusion for the plot title
adj_years = trend_component.index.year[~trend_component.index.year.isin(boundary_years)]

# Scatter plot: mean vs amplitude
plt.figure(figsize=(6,4))
plt.scatter(x, y, alpha=0.6)
plt.plot(x, m*x + b, color='red', linestyle='--', 
         label=f'Slope = {m:.3f}, $r$ = {corr:.2f}\nAvg Amplitude = {avg_amp:.2f} ppb')

plt.xlabel('Annual Mean STL Trend Level (ppb)')
plt.ylabel('Seasonal Amplitude (ppb)')
plt.title(f"Seasonal Amplitude vs. Mean CH4 Level ({adj_years.min()}–{adj_years.max()})")
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```

**Seasonal amplitude vs mean level results summary**

Data Included | Slope | Correlation r | Interpreretation
--------------|-------|---------------|-----------------
All 35 years | 0.050 | 0.47 | Moderate positive association; the linear association accounts for 22% of the observed variation in annual amplitude
Exluding 1991 | 0.063 | 0.61 | Stronger positive association; the linear association accounts for 37% of the observed variation in annual amplitude
Exluding 1991 and 2025 | 0.047 | 0.52 | Positive association remains; the linear association accounts for 27% of the observed variation in annual amplitude


The relationship between mean CH4 concentration and estimated seasonal amplitude is moderately positive across the full record (r = 0.47). Sensitivity analysis shows that excluding 1991 strengthens this association (r = 0.61), while excluding both 1991 and 2025 leaves a positive relationship (r = 0.52). Thus, the association is not solely attributable to either unusual year, although its estimated strength is sensitive to the selected observations. The results support an association between mean CH4 concentration and seasonal amplitude but do not establish causation. The unusually large 1991 amplitude may be sensitive to STL boundary effects, while the lower observational coverage in recent years warrants additional caution in interpreting absolute amplitude estimates.

An important limitation is that mean CH4 trend level is strongly associated with calendar time because of the persistent long-term increase in atmospheric CH4. The amplitude-versus-time analysis also reveals substantial interannual and multiyear variation in seasonal amplitude. Consequently, the observed amplitude–mean association may partly reflect shared temporal structure rather than a direct scaling relationship between CH₄ concentration and seasonal amplitude. This analysis therefore does not establish multiplicative seasonality or, by itself, justify a variance-stabilizing transformation.

## Seasonal structure summary

**Annual CH4 seasonal amplitude and shape summary**

The STL decomposition reveals a remarkably persistent annual CH4 seasonal pattern across 1991–2025. Although the seasonal component varies substantially in magnitude between years, the timing and overall shape of its peaks and troughs remain broadly consistent. Correlations between individual annual seasonal curves and the mean seasonal curve support this interpretation, with 1991 standing out as the clearest exception.

The estimated peak-to-trough seasonal range is particularly large in 1991 and increases progressively from 2022 through 2025. The moderate positive association between seasonal range and mean CH4 concentration suggests that larger mean concentrations may accompany a stronger seasonal cycle, but this relationship does not establish causation. Seasonal methane variability arises from interacting changes in emissions, atmospheric oxidation by hydroxyl radicals (OH), and transport. Recent literature reports increasing methane seasonal-cycle amplitudes in the subtropics and tropics, providing a plausible physical context for the observed pattern [1].

The regular weekly resampling and interpolation provide a consistent time grid for seasonal analysis and forecasting. However, recent observational coverage declines substantially, reaching approximately 60% in 2025. Because some seasonal extrema occur during missing weeks, the estimated recent amplitudes may be sensitive to interpolation. Similarly, the unusually large 1991 estimate warrants caution because it occurs at the beginning of the series and has a substantially different seasonal shape from the long-term average.

Overall, the analysis supports a persistent annual CH₄ seasonal pattern with substantial interannual variation in amplitude. The OCSB test did not recommend seasonal differencing under the specifications examined, although this does not establish that the seasonal pattern is deterministic. The recent increase in estimated amplitude is a noteworthy empirical result that persists under the baseline and wider STL seasonal-smoothing settings, but its magnitude is sensitive to the decomposition specification, and incomplete recent observations limit confidence in the absolute estimates. The amplitude–mean relationship is moderately positive but may partly reflect shared temporal structure rather than direct scaling with CH₄ concentration. For forecasting, annual seasonality should be retained in candidate models; whether a transformation or more flexible seasonal-amplitude structure improves predictions should be determined through training-data diagnostics and out-of-sample validation.

*TL/DR*: 
- Seasonal shape: the annual cycle is remarkably consistent across the 35-year record.
- Seasonal amplitude: the amplitude exhibits substantial interannual variation, including a marked recent increase.
- Reliability: the broad recent increase persists under an alternative STL smoothing setting, but the estimated magnitude is sensitive to that setting, and recent observational coverage is lower.
- Transformation decision: the moderate positive amplitude–mean association is confounded with time and does not establish multiplicative seasonality. Retain the original CH4 scale for the baseline forecasting models and let out-of-sample performance guide any transformation.
